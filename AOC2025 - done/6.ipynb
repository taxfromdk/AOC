```python
import re
def load(fn):
    D = [x.replace('\n','') for x in open(fn).readlines()]
    return D
```

```python
def solve1(fn):
    D=load(fn)
    z=r'[\s,]+'
    zzz=0
    D =[re.split(z,x.strip()) for x in D]
    for i in range(len(D[0])):
        o=D[-1][i]
        v=[D[x][i] for x in range(len(D)-1)]
        #print(v,o)
        if o=='*':
            t=1
            for z in v:
                t*=int(z)
            zzz+=t
        elif o=='+':
            for z in v:
                zzz+=int(z)
    return zzz

print("part1:",solve1("6.txt"), 5060053676136 )

```
