```python
def load(fn):
    M={}
    d=open(fn).readlines()
    w=0
    h=0
    for y,l in enumerate(d):
        for x,v in enumerate(l):
            if v in "@":
                M[x+y*1j]=v
                w=max(w,x)
                h=max(h,y)
        
    return M,w,h        
    
    
def solve1(fn):
    M,w,h=load(fn)
    r=0
    for p in M.keys():
        c=0
        for dp in [-1-1j,-1j,1-1j,-1,1,-1+1j,1j,1+1j]:
            pp=p+dp
            if pp in M:
                c+=1
        if c<4:
            r+=1
    return r

def solve2(fn):
    M,w,h=load(fn)
    r=0
    while 1:
        done=1
        for p in M.keys():
            if not M[p] in "@":
                continue
            c=0
            for dp in [-1-1j,-1j,1-1j,-1,1,-1+1j,1j,1+1j]:
                pp=p+dp
                if pp in M and M[pp] in "@":
                    c+=1
            if c<4:
                r+=1
                done=0
                M[p]='x'
        if done:
            break
    return r

print("part1:",solve1("4.txt"),1464)
print("part2:",solve2("4.txt"),8409)


```
