```python
def load(fn):
    d= open(fn).read()
    r,i=d.split("\n\n")
    r=[sorted([int(z) for z in y]) for y in[x.split("-") for x in r.split("\n")]]
    i=[int(x) for x in i.split("\n")]
    return r,i

def solve1(fn):
    R,I=load(fn)
    g=0
    for i in I:
        for r in R:
            if i>=r[0] and i<=r[1]:
                g+=1
                break
    return g

print("part1:",solve1("5.txt"),652)
```

```python
def solve2(fn):
    R,_=load(fn)
    while(1):
        clean = 1
        for i in range(len(R)):
            for j in range(len(R)):
                if i!=j:
                    a0,a1 = R[i]
                    b0,b1 = R[j]
                    if b1 < a0 or a1 < b0:
                        #no touch
                        pass                    
                    else:
                        R[i] = (min(a0,b0),max(a1,b1))
                        del R[j]
                        clean = 0
                        break
            if not clean:
                break
        
        if clean:
            break    
    return sum([b-a+1 for a,b in R])
    
print("part2:",solve2("5.txt"),341753674214273)
```
