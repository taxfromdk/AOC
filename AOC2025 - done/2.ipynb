```python
def load(fn):
    d=open(fn).read().split(",")
    d=[x.split("-") for x in d]
    d=[(x[0],x[1]) for x in d]
    return d

def increment(x):
    return str(int(float(x))+1)

def invalid(x,part):
    #print(x)
    l=len(x)
    tp=2 if part==1 else l
    #print(x)    
    for parts in range(2,tp+1):
        #
        if l%parts== 0:
            bad=True
            partlen=int(l/parts)
            for j in range(partlen):
                tst=x[j]
                for k in range(1,parts):
                    if tst != x[j+k*partlen]:
                        bad=False
            if bad:
                return True
    return False

def solve(fn, part):
    d=load(fn)
    r=0
    for a,b in d:
        i = a
        while 1:
            if invalid(i,part):
                r+=int(float(i))
            if i == b:
                break
            i=increment(i)
    return r


print("part1:",solve("2.txt",1),29940924880)

print("part2:",solve("2.txt",2), 48631958998  )





```
