```python
def load(fn):
    J=open(fn).readlines()
    J=[(int(a), int(b), int(c)) for a,b,c in [x.split(',') for x in  J]]
    
    
    return J

def solve1(fn, mrg):
    #junctions
    J=load(fn)
    
    def dst(a,b):
        ax,ay,az=a
        bx,by,bz=b
        dx=bx-ax
        dy=by-ay
        dz=bz-az
        return dx*dx+dy*dy+dz*dz
    
    #Distances
    D = {}
    l=len(J)
    for i in range(l):
        for j in range(i+1,l):
            k = (min(i,j),max(i,j))
            D[k] = dst(J[i],J[j])
    
    #sorted distances
    S=[]
    for k in D.keys():
        i,j=k
        S.append((D[k], i, j))
    S = sorted(S, key=lambda x: x[0], reverse=False)
    
    #print(S[:10])
    
    #circuits
    #start by just giving each junction an index
    C = []
    for i in range(l):
        C.append(-i)
    
    
    assert(len(C)==len(J))
    
    #Combine closest connections
    for ix1 in range(mrg):
        _,i,j = S[ix1]
        #print("merging junctions", J[i],J[j])
        #Let all junctions in C[i] be C[j]
        zzz=C[i]
        yyy=C[j]
        for ix2 in range(l):
            if C[ix2] == zzz:
                C[ix2] = yyy
        #print(C)
        
        
    #print(C)
        
    #count remaining circuit ids
    Z = {}
    for x in C:
        if not x in Z:
            Z[x]=0
        Z[x]+=1
        #print(x)
    
    print("Circuits",len(Z.keys()))
    
    a = 1
    for x in sorted(Z.values(),reverse=True)[:3]:
        #print(x)
        a*=x
    
    return a

#low 504
#low 576
#low 13860
#not 305520
#not 13860
#not 285420

print("test1:",solve1("8.tst", 10),40)
print("part1:",solve1("8.txt",1000),103488)
    
```
