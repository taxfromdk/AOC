```python
def load(fn):
    d =open(fn).readlines()
    d=[x.strip() for x in d]
    d=[(x[0],int(x[1:])) for x in d]
    d=[y if x=='R' else -y for x,y in d]
    return d


def solve1(fn):
    d=load(fn)
    r=0
    p=50
    for x in d:
        p+=x
        while p<0:
            p+=100
        while p>=100:
            p-=100
        if p==0:
            r+=1
    
    return r

def solve2(fn):
    
    d=load(fn)
    r=0
    p=50
    for x in d:
        D=1 if x>0 else -1
        X=x*D
        for y in range(X):
            p+=D
            if p==-1:
                p=99
            if p==100:
                p=0
            if p==0:
                r+=1
        
    
    return r

    
    
    
print("part1:",solve1("1.txt"),1177)
print("part2:",solve2("1.txt"),6768)

```
