```python
def mj(b,s):
    l=len(b)
    x=l-(s-1)
    
    best_v=-1
    best_i=None
    for i in range(x):
        v=int(b[i])
        if v>best_v:
            best_v=v
            best_i=i
    
    r = str(best_v)
    if s>1:
        r+=mj(b[best_i+1:],s-1)
    return r
    
def solve(fn,s):
    d=open(fn).readlines()
    d = [x.strip() for x in d]
    r=0
    for b in d:
        r+=int(mj(b,s))
    return r

print("part1:",solve("3.txt",2),17095)
print("part2:",solve("3.txt",12),168794698570517)
```
