```python
def load(fn):
    S=[]
    M={}
    width=0
    height=0
    L=open(fn).readlines()
    for y,l in enumerate(L):
        for x,c in enumerate(l):
            if c in ".^S":
                p=x+y*1j
                if c=='S':
                    S.append(p)
                if c=='^':
                    M[p]=c
                if c=='.':
                    pass
                width=max(width,x+1)
                height=max(height,y+1)
    return M,S,width,height
            
    
    
```

```python
def solve1(fn):
    M,beams,w,h=load(fn)
    #print(beams,w,h)
    splits=0
    #print(M)
    while 1:
        nb=[]
        for b in beams:
            #print(b)
            b+=1j
            if b.imag < h:
                #print(b)
                if b in M:
                    splits+=1
                    if not b+1 in nb:
                        nb.append(b+1)
                    if not b-1 in nb:
                        nb.append(b-1)
                else:
                    if not b in nb:
                        nb.append(b)
        beams=nb
        if len(beams)==0:
            break
    return splits

print("part1:",solve1("7.txt"),1518)
```

```python
def solve2(fn):
    W=load(fn)
    m={}
 
    def rec(b,W):
        nonlocal m
        if b in m:
            return m[b]
        M,beams,w,h=W
        if b.imag < h:
            nb=b
            nb+=1j
            if nb in M:
                r=rec(nb+1,W)+rec(nb-1,W)
            else:
                r=rec(nb,W)
        else:
            r=1
        m[b]=r
        return r
                
    return rec(W[1][0],W)

print("part2:",solve2("7.txt"),25489586715621)
     
     
     
```
