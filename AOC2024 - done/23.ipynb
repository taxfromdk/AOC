```python
def solve1(fn):
    E=[x.strip().split("-") for x in open(fn).readlines()]
    
    #print("E",E)
    
    N = []
    for a,b in E:
        N.append(a)
        N.append(b)
    N=list(set(N))
    
    #print("N", N))
    print("N", len(N), "E", len(E))
    
    
    _E = {}
    for a,b in E:
        if not a in _E:
            _E[a]=[]
        if not b in _E:
            _E[b]=[]
        _E[a].append(b)
        _E[b].append(a)
    E=_E
    #print("E",E)
    
    
    
    r=0
    l=len(N)
    for a in range(l):
        for b in range(a,l):
            for c in range(b,l):
                #different
                if a!=b and b!=c and a!=c:
                    _a = N[a]
                    _b = N[b]
                    _c = N[c]
                    #connected
                    if _a in E[_b] and _b in E[_c] and _c in E[_a]:
                        #print(_a,_b,_c)
                        #could be nisse
                        if _a.startswith("t") or _b.startswith("t") or _c.startswith("t"):
                            r+=1
    return r
            
        

print("test1:",solve1("23.tst"),7)
print("part1:",solve1("23.txt"),1419)




```
