```python
#from functools import cache
#from itertools import combinations
import networkx as nx

    

def solve2(fn):
    E=[x.strip().split("-") for x in open(fn).readlines()]
    
    #print("E",E)
    
    N = []
    for a,b in E:
        N.append(a)
        N.append(b)
    N=list(set(N))
    
    #print("N", N))
    #print("N", len(N), "E", len(E))
    
    
    G = nx.Graph()
    G.add_edges_from(E)
    maximal_cliques = list(nx.find_cliques(G))
    
    # If no cliques found, return empty list
    if not maximal_cliques:
        return []
    
    # Return the largest clique
    return ",".join(sorted(max(maximal_cliques, key=len)))
    
#I spent 24 hours trying to dream up a algorithm only to find 
#myself crashing my machine with exploding memory need.

#Turned to networkx framework and solved within minutes, with runtime elow a second.
    
print("test2:",solve2("23.tst"),"co,de,ka,ta")
print("part2:",solve2("23.txt"),"af,aq,ck,ee,fb,it,kg,of,ol,rt,sc,vk,zh")
                 
```
