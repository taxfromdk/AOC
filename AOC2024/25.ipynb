```python
def solve(fn):
    d=open(fn).read().split("\n\n")
    d=[[l.strip() for l in x.split("\n")] for x in d]
    #print(d[0])
    
    keys=[]
    locks=[]
    
    for s in d:
        z=[0]*5
        for x in range(5):
            for y in range(5):
                if s[y+1][x]=='#':
                    z[x]+=1
        if s[0][0] =='#':
            locks.append(z)
        else:
            keys.append(z)
        #print(z)
    print("locks:",len(locks))
    print("keys:",len(keys))
    m=0
    for i,key in enumerate(keys):
        for j,lock in enumerate(locks):
            good=1
            for k in range(5):
                if lock[k]+key[k]>=5:
                    good=0
                    break
            if good:
                m+=1
                #print(i,j)
                    
                
    
    
    
    return m

print("part1:",solve("25.txt"),1)



```
