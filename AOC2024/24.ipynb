```python
def parse(fn):
    d = open(fn).read().split("\n\n")
    W,G = d
    W=[x.strip().split(": ") for x in W.split("\n")]
    W={x[0]: int(x[1]) for x in W}
    
    G=[x.split(" -> ") for x in G.split("\n")]
    G={b:a.split(" ") for a,b in G}
    return W,G

def solve1(fn):
    W,G=parse(fn)
    #print(W)
    #print(len(G))

    while len(G.keys()):
        for k in G:
            a,o,b = G[k]
            #print(a,o,b)
            if a in W and b in W:
                #print("found")
                if o == 'AND':
                    W[k]=W[a] and W[b]
                elif o == 'OR':
                    W[k]=W[a] or W[b]
                elif o == 'XOR':
                    W[k]=W[a] ^ W[b]
                else:
                    print(o)
                    assert(0)
                del G[k]
                break
    
    v = []
    i = 0
    while 1:
        n = "z%02d"%(i)
        #print(n)
        if n in W:
            v.append(W[n])
        else:
            break
        
        i+=1
    r = 0
    for b in reversed(v):
        r = r << 1
        r+=b
    return r
    

print("test1:", solve1("24.tst"), 4)
print("part1:", solve1("24.txt"), 38869984335432)




```
